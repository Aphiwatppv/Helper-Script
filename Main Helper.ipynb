```python
from servers_config import ServersConfig
```

```python
cfg = ServersConfig.from_file("ServersConfig.xml")

# Print all servers (masking passwords)
print("All servers:")
for s in cfg.all():
    print(s.masked())

# Example: get server by instance + username
print("\nLookup by instance + username:")
server = cfg.get_by_instance_and_username("EBS10", "ebs_admin")
if server:
    print("Found:", server.masked())
else:
    print("No match found")

# Example: flexible search
print("\nSearch all servers in Regensburg with instance=EBS10")
for s in cfg.find(location="Regensburg", instance="EBS10"):
    print(s.masked())
```
